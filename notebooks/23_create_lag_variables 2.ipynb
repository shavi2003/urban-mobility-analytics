```python
import pandas as pd

df = pd.read_csv("../data/mdm2_data_files/master_mobility_model_data.csv")

print(df.shape)
df.head()
```

```python
df["datetime"] = pd.to_datetime(df["datetime"])

df = df.sort_values(["sensor_id", "datetime"])
```

```python
df["ped_lag1"] = df.groupby("sensor_id")["ped"].shift(1)
df["ped_lag24"] = df.groupby("sensor_id")["ped"].shift(24)

df["cyc_lag1"] = df.groupby("sensor_id")["cyc"].shift(1)
df["cyc_lag24"] = df.groupby("sensor_id")["cyc"].shift(24)

df["active_lag1"] = df.groupby("sensor_id")["active"].shift(1)
df["active_lag24"] = df.groupby("sensor_id")["active"].shift(24)
```

```python
print(df[[
    "ped_lag1","ped_lag24",
    "cyc_lag1","cyc_lag24",
    "active_lag1","active_lag24"
]].isna().mean())
```

```python
df = df.dropna(subset=[
    "ped_lag1","ped_lag24",
    "cyc_lag1","cyc_lag24",
    "active_lag1","active_lag24"
])

print(df.shape)
```

```python
out_path = "../data/mdm2_data_files/master_mobility_model_with_lags.csv"

df.to_csv(out_path, index=False)

print("Saved:", out_path)
```
